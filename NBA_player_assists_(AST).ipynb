# My First ML Project



# **Load data**


````python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/nba-player-stats-2019.csv')
df
````

# **Data Preperation**


## **Data seperation as X and Y**

````python
y = df['AST']
y
````

````python
X = df.drop('PTS', axis=1)
X
````

## **Data Splitting**

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=100)
````

````python
X_train
````

````python
X_test
````

# **Model Building**

## **Linear Regression**

#### **Training the model**

````python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

# Drop non-numerical columns from X_train and X_test
X_train_numeric = X_train.drop(['Player', 'Pos', 'Tm'], axis=1)
X_test_numeric = X_test.drop(['Player', 'Pos', 'Tm'], axis=1)

lr.fit(X_train_numeric, y_train)
LinearRegression()
````

### **Applying the model to make a prediction**


````python
y_lr_train_pred = lr.predict(X_train_numeric)
y_lr_test_pred = lr.predict(X_test_numeric)
````

````python
print(y_lr_test_pred)
````

````python
y_lr_test_pred
````

### **Evaluate model performance**

````python
from sklearn.metrics import mean_squared_error, r2_score

lr_train_mse = mean_squared_error(y_train, y_lr_train_pred)
lr_train_r2 = r2_score(y_train, y_lr_train_pred)

lr_test_mse = mean_squared_error(y_test, y_lr_test_pred)
lr_test_r2 = r2_score(y_test, y_lr_test_pred)
````

````python
print('LR MSE (Train): ', lr_train_mse)
print('LR MS2 (Train): ', lr_train_r2)
print('LR MSE (Test): ', lr_test_mse)
print('LR MS2 (Test): ', lr_test_r2)
````

````python
lr_results = pd.DataFrame(['Linear regression', lr_train_mse, lr_train_r2, lr_test_mse, lr_test_r2]).transpose()
lr_results.columns = ['Method', 'Training MSE', 'Training R2', 'Test MSE', 'Test R2']
````

````python
lr_results
````

## **Random Forest**

### **Training the model**

````python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(max_depth=2, random_state=100)
rf.fit(X_train_numeric, y_train)
````

### **Applying the model to make a prediction**

````python
y_rf_train_pred = rf.predict(X_train_numeric)
y_rf_test_pred = rf.predict(X_test_numeric)
````

### **Evaluate model performance**

````python
from sklearn.metrics import mean_squared_error, r2_score

rf_train_mse = mean_squared_error(y_train, y_rf_train_pred)
rf_train_r2 = r2_score(y_train, y_rf_train_pred)

rf_test_mse = mean_squared_error(y_test, y_rf_test_pred)
rf_test_r2 = r2_score(y_test, y_rf_test_pred)
````

````python
rf_results = pd.DataFrame(['Random forest', rf_train_mse, rf_train_r2, rf_test_mse, rf_test_r2]).transpose()
rf_results.columns = ['Method', 'Training MSE', 'Training R2', 'Test MSE', 'Test R2']
rf_results
````

## **Model comparison**

````python
df_models = pd.concat([lr_results, rf_results], axis=0).reset_index(drop=True)
df_models
````
